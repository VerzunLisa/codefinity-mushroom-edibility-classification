## Data Loading & Cleaning

```python
import pandas as pd

df = pd.read_csv('mushrooms.csv')
print(df.head(5))

df['stalk-root'] = df['stalk-root'].str.replace('?', 'u')
df = df.drop_duplicates()
print(df.info())
```

## Exploratory Data Analysis (EDA)

```python
import seaborn as sns
import matplotlib.pyplot as plt

print(df.columns)
sns.countplot(data=df, x='class')
plt.title('class')
plt.show()

sns.countplot(data=df, x='odor', hue='class')
plt.title("odor")
plt.show()

sns.countplot(data=df, x='cap-color', hue='class')
plt.title('cap color')
plt.show()

df_encoded = df.apply(lambda col: pd.factorize(col)[0])

df_corr = abs(df_encoded.corr(method='pearson')['class']).sort_values(ascending=False)[0:12].index.tolist()
print(df_corr)

sns.heatmap(df_encoded[df_corr].corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.show()
```
